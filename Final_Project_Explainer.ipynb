# Technical Explainer: The Geography of Resilience
**Link to Website:** [The Geography of Resilience](https://DrSune.github.io/socialdata2026/)

## 1. Motivation
**The Dataset:** We utilized a primary dataset of 1.5 million Danish residential housing sales (1992-2024), enriched with building characteristics from the BBR (Building and Dwelling Register) and official Energy Performance Certificates (EPC).

**Goal:** Our project investigates the socio-economic and geographic divide in energy resilience. We chose this to move beyond the "average" impact of the energy crisis and show how location (Urban vs. Rural) and individual technology adoption (Solar/Heat Pumps) created a two-tier housing market.

## 2. Basic Stats
**Data Cleaning:** We filtered the 1.5M records to focus on the 2022-2024 crisis era (approx. 226k sales). We handled outliers by removing non-market sales (family transfers) and extreme sqm price deviations. 
**EDA:** Initial exploration showed that while urban houses have higher base prices, rural houses show 3x higher volatility in estimated energy costs.

## 3. Data Analysis
We performed a residual analysis on house prices. We predicted 'expected price' using a simple linear model based on sqm and municipality, then looked at how the 'Energy Label' and 'Heating Type' explained the remaining variance. 
*Key Finding:* A 'G' label in a rural area with a heat pump outperformed an 'F' label with natural gas in terms of value retention.

## 4. Genre
**Genre:** Magazine (Segel & Heer, 4.3). 
**Visual Narrative:** We used a **Martini Glass** structure. 
*   **Stem:** The static density plot establishing the crisis inequality.
*   **Flare:** The interactive scatter plot allowing readers to filter by energy label and investment type.

## 5. Visualizations
**Choice Rationale:** 
*   **Density Plots (KDE):** Chosen to show *risk* (the long tail) rather than just *averages*.
*   **Interactive Scatter:** Necessary to handle the multi-dimensional nature of the trade-off (Energy vs. Price vs. Year).

## 6. Discussion
**Critique:** Our 'Resilience Index' is a proxy based on sale prices and BBR codes. A major limitation is that we do not have actual energy consumption data, only theoretical labels. Future work could integrate the 'Energi Data Service' smart meter data for higher precision.
