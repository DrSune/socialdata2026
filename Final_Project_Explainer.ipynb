# Technical Explainer: The Insulation Tax
**Project:** Social Data Analysis 2026  
**Author:** Frederik Sunesen  
**Website:** [The Insulation Tax](https://DrSune.github.io/socialdata2026/)

## 1. Motivation
The 2022 energy crisis in Denmark created a stark divide in the housing market. Our project investigates the **Insulation Tax**: a structural shift where a building's energy efficiency became a primary driver of price resilience, independent of location. 

**Goal:** To provide a non-scientific reader with a clear narrative on how a macro-economic shock (energy prices) filters through building regulations and geography to affect individual household wealth.

## 2. Basic Stats & Data Integrity
*   **Primary Data:** 1.5 million Danish housing sales (1992-2024) enriched with BBR data (Building and Dwelling Register).
*   **The Proxy Methodology:** Since actual EPC (Energy Performance Certificate) labels were not available for the full 1.5M historical records, we utilized **Building Year** as a robust scientific proxy for energy efficiency. This choice is grounded in the history of **Danish Building Regulations (Bygningsreglementet)**:
    *   **Pre-1961 (G/F):** Buildings from before the first national insulation standards (BR61).
    *   **1961-1977 (E):** Period of basic comfort standards before the 1970s oil crisis triggered major tightening.
    *   **1977-1995 (D):** The 'Oil Crisis Response' era (BR77), characterized by significantly lower U-values for walls and roofs.
    *   **Post-1995 (C-A):** The 'Energy Frame' era (BR95 and later), where holistic energy performance became mandatory.
*   **Preprocessing:** 
    *   Filtered for market-representative sales (Free trade only).
    *   Calculated `sqm_price` to normalize for house size.
*   **Exploratory Insights:** Early EDA confirmed that 'Legacy' buildings (Pre-1977) saw significantly higher 'Price Haggling' (gap between offer and sale) during the crisis compared to modern builds.

## 3. Data Analysis
Our pipeline utilizes three technical layers:
1.  **Temporal Aggregation:** We grouped 1.5M records into Quarterly buckets to isolate the stable pre-crisis (2019) and peak-crisis (2023) signals.
2.  **Regional Growth Analysis:** Calculated the delta in median price growth for legacy vs. modern groups across all 8 administrative areas to prove universality.
3.  **Negotiation Residuals:** We analyzed the absolute percentage change between offer and purchase price to measure 'Market Friction'—a high-signal metric for buyer anxiety during energy volatility.

## 4. Genre & Narrative (Segel & Heer)
Our project strictly follows the **Magazine Style** genre with a **Martini Glass** structure.

### Visual Narrative Tools
*   **Visual Structuring:** We use explicit headings and a linear scrolling layout to lead the reader through the 'Stem' of the narrative.
*   **Highlighting:** We employ a consistent semantic color palette (Red = Legacy/Taxed, Green = Modern/Resilient) across all figures to maintain continuity.
*   **Transition Guidance:** Each visualization is preceded by introductory text that frames the next question.

### Narrative Structure Tools
*   **Ordering:** Linear ordering for the first two figures (author-driven) to establish the facts, opening into user-directed ordering in Figure 3.
*   **Interactivity:** We progress from static (author-driven) to fully interactive (Exploratory Scatter with hover and filtering).
*   **Messaging:** We use 'Captions' to explicitly coach the reader on the takeaway of each plot, following the 'No Black Box' philosophy.

## 5. Visualizations
1.  **Ridgeline Plots (Seaborn):** Chosen to show the authentic 'Decoupling' of price distributions. Unlike a bar chart of means, ridges show how the entire price floor for legacy homes shifted left.
2.  **Regional Facet Bars (Matplotlib/Seaborn):** A 'Small Multiples' approach that proves the Insulation Tax isn't just a rural issue—it is universal across Denmark.
3.  **Interactive Regression Scatter (Plotly):** The 'Flare'. This allows the reader to hover over 2000 real sales from 2023 and see the exact 'Haggling %' for individual houses across the age-spectrum.

## 6. Discussion
**Successes:** Using 'Building Year' as a proxy allowed us to analyze 1.5 million sales with high historical depth. The resulting 'Decoupling' signal (Figure 1) is extremely strong and aligns perfectly with the timing of the energy crisis.

**Limitations:** While Building Year is a strong proxy, individual home renovations (e.g., retrofitted heat pumps) can improve an old home's performance. This explains the outliers in Figure 3 who maintain high prices despite being 'Legacy' builds.
