# Technical Explainer: The Insulation Tax
**Project:** Social Data Analysis 2026  
**Author:** Frederik Sunesen  
**Website:** [The Insulation Tax](https://DrSune.github.io/socialdata2026/)

## 1. Motivation & Accessing Data
The 2022 energy crisis in Denmark served as a macro-economic catalyst that fundamentally re-indexed the value of national housing stock. Our objective was to move beyond simple price monitoring to investigate **The Insulation Tax**: a structural shift where a building's technical performance (energy efficiency) replaced location as the primary predictor of price resilience.

### Learning Objective Alignment:
*   **Access & Assess Data:** We accessed 1.5 million historical sales records (1992-2024) from online sources, assessing their suitability for a multi-decade temporal analysis.
*   **Goal:** To communicate complex spatial-temporal economic shifts to a non-scientific audience through a cohesive data narrative.

## 2. Data Sources
Transparency and reproducibility are core to our methodology. The following datasets are the actual files used in the analysis:
*   **Primary Dataset (1.5M Records):** [full_housing_data.parquet](https://github.com/DrSune/socialdata2026/raw/main/full_housing_data.parquet) - The raw historical sales foundation (42MB).
*   **Processed Visualization Data:** [final_viz_data.csv](https://github.com/DrSune/socialdata2026/raw/main/final_viz_data.csv) - The enriched dataset including geocoding and energy era proxies.
*   **External Regulatory Context:** Historical mandates accessed via [Bygningsreglementet.dk](https://bygningsreglementet.dk/Historik), used to define our scientific 'Energy Eras'.
*   **Address Normalization:** Data retrieved via the [DAWA API](https://api.dataforsyningen.dk) to map municipal boundaries.

## 3. Filtering, Cleaning & Basic Stats
Handling a 1.5 million record dataset required state-of-the-art Python tools (`pandas`, `numpy`, `pyarrow`).
### Preprocessing Steps:
*   **Filtering:** Isolated 'Free Trade' sales to exclude family transfers and ensure market-representative pricing.
*   **Organization:** Grouped data by 'Area' (8 administrative regions) and 'Efficiency Eras'.
*   **Proxy Methodology:** Since direct EPC labels were unavailable for historical data, we utilized **Danish Building Regulations (BR61, BR77, BR95)** as a robust scientific proxy for efficiency based on construction year.

## 4. Data Analysis & Exploratory Visualization
We applied the full spectrum of course techniques to validate our hypothesis:
*   **Distribution Analysis (Single Variable):** Using **Kernel Density Estimates (KDE)** in Figure 1, we identified the 'Great Decoupling'—a shift in the entire price floor that would be hidden by simple summary statistics.
*   **Functional Relationships (Multiple Variables):** In Figure 3, we estimated the functional relationship between building age and price using **OLS Regression** and **smoothing noise** through sampling.
*   **Limitations of Summary Statistics:** By plotting the full ridgeline distributions, we addressed the limitations of mean/median reporting, proving that the 'Penalty' isn't just a lower average, but a fundamentally different distribution shape.

## 5. Genre & Narrative (Segel & Heer Analysis)
Our project is structured as a **Magazine Style** narrative using a **Martini Glass** architecture, as defined in the seminal paper *Narrative Visualization: Telling Stories with Data* by Segel & Heer (2010).

### Narrative Structuring (Segel & Heer Framework):
Following the 'Design Space' established by Segel & Heer, we utilize specific tools across their three core categories:
*   **Visual Narrative Tools:** We use *Visual Structuring* through consistent headers and a linear scrolling path. We also employ *Highlighting* via a semantic color palette (Red/Green) that persists across all figures.
*   **Narrative Structure Tools:** We follow a *Linear Ordering* for the 'Stem' of the story (Figures 1-2) before opening into a *User-Directed* 'Flare' (Figure 3). *Messaging* is provided through coaching captions for every plot.
*   **Interactivity Tools:** We move from static snapshots to *Filtering* and *Hover* interactions in the final interactive scatter plot.

## 6. AI Disclosure & Ethical Use
As per modern academic standards and the 'No Black Box' course policy, we explicitly disclose the use of AI in this project:
*   **Development:** Large Language Models (LLMs) were used as coding assistants to help write modular Python scripts for data cleaning and visualization generation.
*   **Drafting:** AI was used to help structure the 'Magazine' narrative and ensure consistency in the Tufte-style graphical integrity pass.
*   **Integrity:** All final data engineering, statistical modeling (OLS), and regulatory proxying were verified against official Danish Building Regulations and the 1.5M record housing dataset to ensure scientific accuracy.

## 8. Discussion & Critical Reflection
**What went well:** The use of Building Regulations as a proxy for efficiency proved highly successful. It allowed us to reveal a clear 'Decoupling' (Figure 1) that aligns perfectly with the 2022 energy price spike. The 'Haggling %' metric (Negotiation Residuals) was a breakthrough, providing a much faster-reacting market signal than final sale prices.

**What is still missing:** Our analysis assumes that construction year perfectly correlates with current insulation levels. While we addressed 'Renovated Outliers' in the narrative, a direct dataset of EPC (Energy Performance Certificate) labels for all 1.5M records would provide higher precision.

**What could be improved:** Future iterations could include a machine-learning model (e.g., Random Forest) to predict sale prices based on energy labels, city, and size, allowing us to generate a 'Predicted Insulation Tax' for any given address. This was out of scope for the current standalone narrative but remains a logical next step.
