# Technical Explainer: The Insulation Tax
**Project:** Social Data Analysis 2026  
**Author:** Frederik Sunesen  
**Website:** [The Insulation Tax](https://DrSune.github.io/socialdata2026/)

## 1. Motivation
The 2022 energy crisis in Denmark was a unique economic event that suddenly re-indexed the value of the Danish housing stock. Our project moves beyond basic price monitoring to investigate **The Insulation Tax**: a structural shift where a building's technical performance (efficiency) replaced geography as the primary predictor of price resilience.

**Goal:** We aimed to create a standalone investigative article that explains this paradigm shift to a non-scientific audience, using 1.5 million real-world data points to prove that the 'Location, Location, Location' mantra has been fundamentally challenged.

## 2. Basic Stats & Data Integrity
*   **Primary Data:** 1.5 million Danish housing sales (1992-2024) enriched with BBR data (Building and Dwelling Register).
*   **The Proxy Methodology:** Since actual EPC (Energy Performance Certificate) labels were not available for the full historical dataset, we utilized **Building Year** as a robust scientific proxy for energy efficiency. This choice is grounded in the history of **Danish Building Regulations (Bygningsreglementet)**:
    *   **Pre-1961 (G/F):** Buildings constructed before the first mandatory thermal insulation standards (BR61).
    *   **1961-1977 (E):** Basic requirements that were drastically tightened following the **1973 Oil Crisis**.
    *   **1977-1995 (D):** The BR77 era, which halved allowed U-values for walls and roofs.
    *   **Post-1995 (C-A):** Modern standards following the revolutionary BR95 'Energy Frame' paradigm.
*   **Preprocessing:** We filtered for market-representative 'Free Trade' sales and calculated `sqm_price` to control for house size variation.

## 3. Data Analysis
Our pipeline utilizes three technical layers:
1.  **Distribution Analysis:** We analyzed the entire price spectrum using kernel density estimations to observe how the *floor* of the market moved relative to the *ceiling*.
2.  **Aggregation by Administrative Area:** We grouped 1.5M records into 8 regions to prove that the 'Insulation Tax' was not a localized phenomenon but a national one.
3.  **Negotiation Residuals:** We analyzed the percentage change between offer and purchase price to measure **Market Friction** (buyer anxiety), a metric that reacts much faster than final indices.

## 4. Genre & Narrative (Segel & Heer)
We strictly follow the **Magazine Style** genre with a **Martini Glass** structure.

### Visual Narrative Tools
*   **Visual Structuring:** Explicit headings and a large-format typography guide the reader through the article.
*   **Highlighting:** A consistent semantic color palette (Red = Legacy/Penalty, Green = Modern/Resilience) is used across all plots.
*   **Transition Guidance:** Introductory paragraphs frame each visualization as an answer to a specific question posed by the previous section.

### Narrative Structure Tools
*   **Ordering:** Linear ordering for Figures 1 and 2 (author-driven facts) leading into Figure 3 (user-driven exploration).
*   **Interactivity:** We move from static snapshots (Ridges) to a fully interactive 'Flare' (Explorer Scatter) where the reader can hover over individual house sales.
*   **Messaging:** Captions follow the 'No Black Box' rule by explaining the takeaway of the data (e.g., 'Modern homes outperformed legacy homes everywhere').

## 5. Visualizations
1.  **Figure 1 (Ridgeline Plots):** Chosen for its high signal-to-noise ratio in showing the authentic 'Decoupling' of price distributions over time.
2.  **Figure 2 (Small Multiples):** Provides a comprehensive national audit. It demonstrates that geography cannot shield an inefficient asset from the Insulation Tax.
3.  **Figure 3 (Interactive Explorer):** The 'Flare'. This allows the reader to explore the micro-reality of 2023. By mapping age against price with bubble size representing 'Haggling', it visualizes the physical slope of the energy crisis.

## 6. Discussion
**Reflection:** The 'Haggling' metric proved to be the most surprising insight. It showed that even if an old home eventually sold for a high price, the *anxiety* (negotiation pressure) was significantly higher than for modern builds.

**Technical Integrity:** By using regulatory eras (BR61-BR95) rather than arbitrary decades, we've anchored the project in the real physics of the Danish building stock.
