# Technical Explainer: The Green Buffer
**Project:** Social Data Analysis 2026  
**Author:** Frederik Sunesen  
**Website:** [The Green Buffer](https://DrSune.github.io/socialdata2026/)

## 1. Motivation
The 2022 energy crisis in Denmark created a stark divide in the housing market. Our project moves beyond simple price trends to investigate a unique hypothesis: **Can proximity to nature act as a financial buffer against the 'Insulation Penalty'?** We chose this topic to explore how environmental amenities interact with economic shocks in a spatial context.

## 2. Basic Stats
*   **Primary Data:** 1.5 million Danish housing sales (1992-2024) with BBR characteristics (building year, address, price).
*   **Nature Data:** Protected nature polygons (§3 Naturbeskyttelsesloven) from the Danish Environmental Portal (GeoJSON).
*   **Cleaning:** We filtered for market sales (excluding family transfers) and focused on the 2019 (Stable) vs 2023 (Crisis) periods for our comparative analysis.

## 3. Data Analysis & Methodology
Our pipeline consists of four key technical stages:
1.  **Geocoding:** We used the **DAWA (Danish Address Web Architecture)** API to convert street addresses into precise coordinates (lat/lon).
2.  **Spatial Analysis:** Using `geopandas` and `shapely`, we projected all data to **UTM 32N (EPSG:25832)**. We then calculated the minimum Euclidean distance between each house and the nearest protected nature polygon.
3.  **Residual Analysis:** To isolate the 'Nature Premium', we calculated price residuals—the percentage deviation of a house's price from its **municipal median** for that year. This controls for local market variations.
4.  **Binned Categorization:** We categorized nature proximity into four levels: Immediate (<1km), Close (1-5km), Nearby (5-10km), and Distant (>10km).

## 4. Genre & Narrative
**Genre:** Magazine. We follow a **Martini Glass** structure:
*   **The Stem:** A static slope chart establishing the macro-economic trend (The Insulation Penalty).
*   **The Pivot:** A geographic map showing the spatial distribution of 'Resilient' houses (those near nature).
*   **The Flare:** An interactive strip plot allowing readers to explore individual sales and verify the 'Green Buffer' hypothesis themselves.

## 5. Visualizations
*   **Slope Chart:** Chosen for its high signal-to-noise ratio in showing divergence over time. It clearly illustrates the 'Insulation Penalty' widening between 2019 and 2023.
*   **Folium Map:** Employs a 'Resilient Outliers' methodology. Instead of muddy heatmaps or dense clusters, we isolate the anomaly: older houses (G-D labels) near nature that sold *above* their expected local median. Background sales are kept small and transparent to highlight the high-signal green outliers.
*   **Plotly Box-and-Strip Plot:** Allows multi-dimensional exploration (Efficiency vs. Residuals vs. Nature). By combining a box plot (for median and quartile clarity) with all data points (`points='all'`), it avoids the overplotting issues of a standard scatter plot while retaining the ability to hover over individual sales.

## 6. Discussion
**Limitations:** Our nature proximity is based on distance to any protected area, regardless of type (forest vs. marsh). Future iterations could weight these differently. Furthermore, our energy labels are proxies based on building year, as actual EPC certificates were not available for the full 1.5M dataset.
