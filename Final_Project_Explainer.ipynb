# Technical Explainer: The Market Decoupling
**Project:** Social Data Analysis 2026  
**Author:** Frederik Sunesen  
**Website:** [The Market Decoupling](https://DrSune.github.io/socialdata2026/)

## 1. Motivation
The 2022 energy crisis in Denmark created a stark divide in the housing market. Our project investigates the **Market Decoupling**: a structural shift where a building's energy efficiency (insulation) became a primary driver of price resilience, independent of location. 

**Goal:** We wanted to provide a non-scientific reader with a clear narrative on how a macro-economic shock (energy prices) filters through geography and building age to affect individual household wealth.

## 2. Basic Stats
*   **Primary Data:** 1.5 million Danish housing sales (1992-2024) enriched with BBR data.
*   **Preprocessing:** 
    *   Filtered for market-representative sales (Free trade only).
    *   Calculated `sqm_price` to normalize for house size.
    *   Proxied 'Energy Efficiency' eras based on Danish building regulation history (Pre-1961, 1961-77, etc.).
*   **Exploratory Insights:** Early EDA showed that while the whole market dipped in 2022, the 'Legacy' building stock (G/F labels) experienced significantly higher 'Price Haggling' (gap between offer and sale) compared to modern high-efficiency builds.

## 3. Data Analysis
Our pipeline utilizes three technical layers:
1.  **Temporal Aggregation:** We grouped 1.5M records into Quarterly and Monthly buckets to isolate the pre-crisis (2019) and peak-crisis (2023) signals.
2.  **Geospatial Processing:** Using administrative area mappings to track the regional distribution of the price penalty.
3.  **Haggling Residuals:** We calculated the absolute percentage change between offer and purchase price to measure 'Market Friction'—a high-signal metric for buyer anxiety.

## 4. Genre & Narrative (Segel & Heer)
Our project strictly follows the **Magazine Style** genre with a **Martini Glass** structure.

### Visual Narrative Tools
*   **Visual Structuring:** We use explicit headings and a linear scrolling layout to lead the reader through the 'Stem' of the narrative.
*   **Highlighting:** We employ a consistent semantic color palette (Red = Price Stress/Legacy, Green/Blue = Resilience/Modern) across all three figures to maintain continuity.
*   **Transition Guidance:** Each visualization is preceded by introductory text that frames the next question (e.g., transitioning from macro trends to geographic vulnerability).

### Narrative Structure Tools
*   **Ordering:** Linear ordering for the first two figures (author-driven) to establish the facts, opening into user-directed ordering in Figure 3.
*   **Interactivity:** We progress from static (author-driven) to semi-interactive (Temporal Slider) to fully interactive (Filtering in Parallel Categories).
*   **Messaging:** We use 'Captions' to explicitly coach the reader on the takeaway of each plot, following the 'No Black Box' philosophy.

## 5. Visualizations
1.  **Ridgeline Plots (Seaborn):** Chosen to show the authentic 'Decoupling' of price distributions. Unlike a bar chart of means, ridges show how the entire price floor for legacy homes shifted left.
2.  **Animated Bubble Map (Plotly):** Provides the 'Pivot'. By scrubbing the timeline, the reader experiences the temporal onset of the crisis across Denmark's regions.
3.  **Parallel Categories Flow (Plotly):** The 'Flare'. This multi-variate exploratory tool lets the reader test their own hypotheses about how Era, Region, and Price Stress interact.

## 6. Discussion
**Successes:** The transition from author-driven facts to reader-driven exploration creates a satisfying narrative arc. The 'Haggling' metric proved to be a much more sensitive indicator of the energy crisis than final sale price alone.

**Improvements:** Future versions could integrate local heating source data (e.g., District Heating vs. Individual Gas) to further refine the 'Vulnerability' map.
